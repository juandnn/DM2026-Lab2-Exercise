### 1. Data Loading with the train/validation/test split.


````python

import pandas as pd
from sklearn.model_selection import train_test_split


def load_and_split_data(
    data_dir="/kaggle/input/competitions/2026-dm-lab-2-private-competition",
    val_size=0.20,
    random_state=42
):
    """
    Load, merge, and split the emotion classification dataset.

    Returns:
        train_df: 80% of original training data
        val_df:   20% of original training data
        test_df:  Original competition test data

    Each DataFrame contains: post_id, text, label.
    """

    # 1. Load datasets
    identification = pd.read_csv(f"{data_dir}/data_identification.csv")
    emotions = pd.read_csv(f"{data_dir}/emotion.csv")
    posts = pd.read_json(f"{data_dir}/final_posts.json")

    # 2. Validate unique post IDs
    for name, df in [
        ("identification", identification),
        ("emotions", emotions),
        ("posts", posts)
    ]:
        if df["post_id"].duplicated().any():
            raise ValueError(f"Duplicate post_id found in {name}")

    # 3. Merge datasets using post_id
    df = identification.merge(
        posts[["post_id", "text"]],
        on="post_id",
        how="left",
        validate="one_to_one"
    )

    df = df.merge(
        emotions[["post_id", "label"]],
        on="post_id",
        how="left",
        validate="one_to_one"
    )

    # 4. Validate data integrity
    if df["text"].isna().any():
        raise ValueError("Some posts are missing text.")

    if not df["set"].isin(["train", "test"]).all():
        raise ValueError("Unexpected dataset assignment found.")

    # 5. Separate original train and test sets
    original_train = df[df["set"] == "train"].copy()
    test_df = df[df["set"] == "test"].copy()

    if original_train["label"].isna().any():
        raise ValueError("Training data contains missing labels.")

    # 6. Stratified 80/20 split of original training data
    train_df, val_df = train_test_split(
        original_train,
        test_size=val_size,
        stratify=original_train["label"],
        random_state=random_state
    )

    # 7. Keep only the required columns
    columns = ["post_id", "text", "label"]

    train_df = train_df[columns].reset_index(drop=True)
    val_df = val_df[columns].reset_index(drop=True)
    test_df = test_df[columns].reset_index(drop=True)

    # 8. Verify no overlapping post IDs
    train_ids = set(train_df["post_id"])
    val_ids = set(val_df["post_id"])
    test_ids = set(test_df["post_id"])

    assert train_ids.isdisjoint(val_ids), "Train/validation overlap!"
    assert train_ids.isdisjoint(test_ids), "Train/test overlap!"
    assert val_ids.isdisjoint(test_ids), "Validation/test overlap!"

    assert len(train_df) + len(val_df) == len(original_train)
    assert len(test_df) == len(df[df["set"] == "test"])

    # 9. Display dataset information
    print("Dataset shapes:")
    print(f"Train:      {train_df.shape}")
    print(f"Validation: {val_df.shape}")
    print(f"Test:       {test_df.shape}")

    print("\nEmotion distributions:")
    for name, split in [
        ("Train", train_df),
        ("Validation", val_df),
        ("Test", test_df)
    ]:
        print(f"\n{name}:")
        distribution = pd.DataFrame({
            "count": split["label"].value_counts(dropna=False),
            "percentage": split["label"].value_counts(
                normalize=True, dropna=False
            ).mul(100).round(2)
        })
        print(distribution)

    print("\nNo post_id overlap between splits: VERIFIED")

    return train_df, val_df, test_df
````

````python

import pandas as pd


def validate_datasets(train_df, val_df, test_df):
    """
    Validate emotion classification datasets.

    Checks:
        1. Missing values
        2. Duplicate post IDs and (text, label) combinations
        3. Valid emotion labels and class distributions
        4. Cross-dataset post ID and text overlaps

    Does not modify the input DataFrames.
    """

    valid_labels = {
        "anger", "disgust", "fear",
        "sadness", "surprise", "joy"
    }

    datasets = {
        "Train": train_df,
        "Validation": val_df,
        "Test": test_df
    }

    print("=" * 65)
    print("DATASET VALIDATION REPORT")
    print("=" * 65)

    # --------------------------------------------------
    # 1. Missing values
    # --------------------------------------------------
    print("\n[1] MISSING VALUES")
    print("-" * 65)

    for name, df in datasets.items():
        print(f"\n{name} dataset ({len(df):,} rows)")

        missing = pd.DataFrame({
            "Count": df.isna().sum(),
            "Percentage (%)": (
                df.isna().mean() * 100
            ).round(2)
        })

        print(missing.to_string())

        for column in df.columns:
            count = df[column].isna().sum()

            if count > 0:
                if name == "Test" and column == "label":
                    print(
                        f"  INFO: {count:,} missing test labels "
                        "(expected for competition submissions)."
                    )
                else:
                    print(
                        f"  WARNING: {count:,} missing values "
                        f"in '{column}'."
                    )

    # --------------------------------------------------
    # 2. Duplicates
    # --------------------------------------------------
    print("\n[2] DUPLICATE ANALYSIS")
    print("-" * 65)

    for name, df in datasets.items():
        print(f"\n{name} dataset")

        n = len(df)

        duplicate_ids = df.duplicated(
            subset=["post_id"]
        ).sum()

        duplicate_text_label = df.duplicated(
            subset=["text", "label"]
        ).sum()

        results = pd.DataFrame({
            "Duplicate type": [
                "post_id",
                "(text, label)"
            ],
            "Count": [
                duplicate_ids,
                duplicate_text_label
            ],
            "Percentage (%)": [
                round(duplicate_ids / n * 100, 2) if n else 0,
                round(duplicate_text_label / n * 100, 2) if n else 0
            ]
        })

        print(results.to_string(index=False))

        if duplicate_ids > 0:
            print("  WARNING: Duplicate post_id values detected.")

        if duplicate_text_label > 0:
            print(
                "  WARNING: Duplicate (text, label) "
                "combinations detected."
            )

        if duplicate_ids == 0 and duplicate_text_label == 0:
            print("  OK: No duplicates detected.")

    # --------------------------------------------------
    # 3. Label validation
    # --------------------------------------------------
    print("\n[3] LABEL VALIDATION")
    print("-" * 65)

    for name, df in datasets.items():
        print(f"\n{name} dataset")

        non_null_labels = df["label"].dropna()

        invalid_labels = non_null_labels[
            ~non_null_labels.isin(valid_labels)
        ]

        if invalid_labels.empty:
            print("  OK: All non-null labels are valid.")
        else:
            print(
                f"  WARNING: {len(invalid_labels):,} "
                "rows contain invalid labels."
            )
            print("\nInvalid label counts:")
            print(
                invalid_labels.value_counts(
                    dropna=False
                ).to_string()
            )

        if name in ["Train", "Validation"]:
            print("\nEmotion distribution:")

            distribution = pd.DataFrame({
                "Count": df["label"].value_counts(),
                "Percentage (%)": (
                    df["label"].value_counts(normalize=True) * 100
                ).round(2)
            })

            print(distribution.to_string())

            missing_classes = valid_labels - set(non_null_labels)

            if missing_classes:
                print(
                    "  WARNING: Missing emotion classes: "
                    f"{sorted(missing_classes)}"
                )
            else:
                print("  OK: All six emotion classes are present.")

        else:
            print(
                f"  INFO: {df['label'].isna().sum():,} "
                "test rows have no label."
            )

    # --------------------------------------------------
    # 4. Data leakage across datasets
    # --------------------------------------------------
    print("\n[4] CROSS-DATASET LEAKAGE")
    print("-" * 65)

    pairs = [
        ("Train", "Validation"),
        ("Train", "Test"),
        ("Validation", "Test")
    ]

    for name_a, name_b in pairs:
        df_a = datasets[name_a]
        df_b = datasets[name_b]

        print(f"\n{name_a} vs {name_b}")

        # Compare post IDs, ignoring missing values
        ids_a = set(df_a["post_id"].dropna())
        ids_b = set(df_b["post_id"].dropna())

        overlapping_ids = ids_a.intersection(ids_b)

        # Compare exact text strings, ignoring missing values
        texts_a = set(df_a["text"].dropna())
        texts_b = set(df_b["text"].dropna())

        overlapping_texts = texts_a.intersection(texts_b)

        print(
            f"  Overlapping post_id values: "
            f"{len(overlapping_ids):,}"
        )
        print(
            f"  Overlapping exact texts:   "
            f"{len(overlapping_texts):,}"
        )

        if overlapping_ids:
            print(
                "  WARNING: Shared post IDs detected "
                "between datasets."
            )

        if overlapping_texts:
            print(
                "  WARNING: Identical texts detected "
                "between datasets."
            )

        if not overlapping_ids and not overlapping_texts:
            print("  OK: No cross-dataset overlaps detected.")

    print("\n" + "=" * 65)
    print("DATASET VALIDATION COMPLETED")
    print("=" * 65)

````

````python

def solve_validation_problems(train_df, val_df):
    """
    Remove (text, label) duplicates and train-validation text leakage.
    Training data is always prioritized over validation data.
    """
    # 1. Remove duplicates within each dataset
    train_df_clean = train_df.drop_duplicates(
        subset=["text", "label"], keep="first"
    ).copy()

    val_df_clean = val_df.drop_duplicates(
        subset=["text", "label"], keep="first"
    ).copy()

    print("Duplicate removal:")
    print(f"  Train: {len(train_df) - len(train_df_clean)} rows removed")
    print(f"  Val:   {len(val_df) - len(val_df_clean)} rows removed")

    # 2. Remove validation texts already present in training
    before = len(val_df_clean)

    val_df_clean = val_df_clean[
        ~val_df_clean["text"].isin(train_df_clean["text"])
    ].copy()

    print("\nTrain-validation leakage removal:")
    print(f"  Train: 0 rows removed (prioritized)")
    print(f"  Val:   {before - len(val_df_clean)} rows removed")

    # 3. Reset indices
    train_df_clean = train_df_clean.reset_index(drop=True)
    val_df_clean = val_df_clean.reset_index(drop=True)

    print("\nFinal dataset shapes:")
    print(f"  Train: {train_df_clean.shape}")
    print(f"  Val:   {val_df_clean.shape}")

    return train_df_clean, val_df_clean
````

````python

import re

def preprocessing_core(text):
    """
    Basic text cleaning: lowercase, remove URLs and mentions,
    and normalize whitespace.
    """
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    return text
````

### 1. Bag-of-Words (BoW)


````python
from sklearn.feature_extraction.text import CountVectorizer


def bag_of_words(
    train_df,
    val_df,
    test_df,
    ngram_range=(1, 2),
    max_features=10000,
    min_df=2,
    max_df=0.95,
    binary=False
):
    """
    Extract Bag-of-Words features from lemmatized tokens.
    Fit only on training data to prevent data leakage.
    """
    vectorizer = CountVectorizer(
        tokenizer=lambda tokens: tokens,
        preprocessor=lambda tokens: tokens,
        token_pattern=None,
        lowercase=False,
        ngram_range=ngram_range,
        max_features=max_features,
        min_df=min_df,
        max_df=max_df,
        binary=binary
    )

    # Fit on training data only
    X_train_bow = vectorizer.fit_transform(train_df["tokens_lemma"])

    # Transform validation and test using the same vocabulary
    X_val_bow = vectorizer.transform(val_df["tokens_lemma"])
    X_test_bow = vectorizer.transform(test_df["tokens_lemma"])

    print("Bag-of-Words feature extraction:")
    print(f"  Train shape:      {X_train_bow.shape}")
    print(f"  Validation shape: {X_val_bow.shape}")
    print(f"  Test shape:       {X_test_bow.shape}")
    print(f"  Vocabulary size:  {len(vectorizer.vocabulary_):,}")

    return X_train_bow, X_val_bow, X_test_bow, vectorizer
````

### 2. TF-IDF

````python
from sklearn.feature_extraction.text import TfidfVectorizer


def tfidf_vectorization(
    train_df,
    val_df,
    test_df,
    ngram_range=(1, 2),
    max_features=10000,
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
):
    """
    Extract TF-IDF features from lemmatized tokens.
    Fit only on training data to prevent data leakage.
    """
    vectorizer = TfidfVectorizer(
        tokenizer=lambda tokens: tokens,
        preprocessor=lambda tokens: tokens,
        token_pattern=None,
        lowercase=False,
        ngram_range=ngram_range,
        max_features=max_features,
        min_df=min_df,
        max_df=max_df,
        sublinear_tf=sublinear_tf
    )

    # Fit on training data only
    X_train_tfidf = vectorizer.fit_transform(train_df["tokens_lemma"])

    # Transform validation and test using the same vocabulary and IDF weights
    X_val_tfidf = vectorizer.transform(val_df["tokens_lemma"])
    X_test_tfidf = vectorizer.transform(test_df["tokens_lemma"])

    print("TF-IDF feature extraction:")
    print(f"  Train shape:      {X_train_tfidf.shape}")
    print(f"  Validation shape: {X_val_tfidf.shape}")
    print(f"  Test shape:       {X_test_tfidf.shape}")
    print(f"  Vocabulary size:  {len(vectorizer.vocabulary_):,}")

    return X_train_tfidf, X_val_tfidf, X_test_tfidf, vectorizer
````

### 3. FastText

````python
import numpy as np
from gensim.models import FastText


def fasttext_vectorization(
    train_df,
    val_df,
    test_df,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    min_n=3,
    max_n=6,
    epochs=30,
    negative=5,
    workers=1,
    seed=42
):
    """
    Train FastText on training tokens only and generate
    document embeddings using mean pooling.
    """
    # 1. Train FastText exclusively on training data
    fasttext_model = FastText(
        sentences=train_df["tokens"].tolist(),
        vector_size=vector_size,
        window=window,
        min_count=min_count,
        sg=sg,
        min_n=min_n,
        max_n=max_n,
        epochs=epochs,
        negative=negative,
        workers=workers,
        seed=seed
    )

    # 2. Convert each document into its average word embedding
    def document_vector(tokens):
        vectors = [
            fasttext_model.wv[token]
            for token in tokens
            if token and token in fasttext_model.wv
        ]
        return (
            np.mean(vectors, axis=0)
            if vectors else np.zeros(vector_size, dtype=np.float32)
        )

    # 3. Transform all datasets using the same trained model
    def transform(df):
        return np.vstack(
            df["tokens"].apply(document_vector).to_numpy()
        ).astype(np.float32)

    X_train_fasttext = transform(train_df)
    X_val_fasttext = transform(val_df)
    X_test_fasttext = transform(test_df)

    # 4. Display results
    print("FastText feature extraction:")
    print(f"  Train shape:      {X_train_fasttext.shape}")
    print(f"  Validation shape: {X_val_fasttext.shape}")
    print(f"  Test shape:       {X_test_fasttext.shape}")
    print(f"  Vocabulary size:  {len(fasttext_model.wv.key_to_index):,}")

    return (
        X_train_fasttext,
        X_val_fasttext,
        X_test_fasttext,
        fasttext_model
    )
````

### 4. MiniLM

````python

import numpy as np
import torch
from sentence_transformers import SentenceTransformer


def minilm_vectorization(train_df, val_df, test_df):
    """
    Generate normalized 384-dimensional MiniLM sentence embeddings.
    Uses preprocessed text directly without fine-tuning.
    """
    # 1. Select device and load pretrained model
    device = "cuda" if torch.cuda.is_available() else "cpu"
    minilm_model = SentenceTransformer(
        "sentence-transformers/all-MiniLM-L6-v2",
        device=device
    )

    # 2. Encode each dataset independently
    def transform(df):
        return minilm_model.encode(
            df["text"].tolist(),
            batch_size=32,
            normalize_embeddings=True,
            convert_to_numpy=True,
            show_progress_bar=True,
            precision="float32",
            truncate_dim=None
        ).reshape(-1, 384).astype(np.float32)

    X_train_minilm = transform(train_df)
    X_val_minilm = transform(val_df)
    X_test_minilm = transform(test_df)

    # 3. Display results
    print(f"MiniLM device: {device}")
    print(f"  Train shape:      {X_train_minilm.shape}")
    print(f"  Validation shape: {X_val_minilm.shape}")
    print(f"  Test shape:       {X_test_minilm.shape}")

    return X_train_minilm, X_val_minilm, X_test_minilm, minilm_model
````

### 5. Dictionary with the transformations

````python

def generate_transformations(train_df, val_df, test_df):
    """
    Generate 12 feature extraction configurations.
    Returns a dictionary containing train, val, test, and info.
    """

    # 1. Configuration definitions
    configs = {
        "BoW-1": ("BoW", dict(ngram_range=(1, 1), min_df=2, binary=False)),
        "BoW-2": ("BoW", dict(ngram_range=(1, 2), min_df=2, binary=False)),
        "BoW-3": ("BoW", dict(ngram_range=(1, 2), min_df=2, binary=True)),

        "TF-IDF-1": ("TF-IDF", dict(ngram_range=(1, 1), min_df=2, sublinear_tf=False)),
        "TF-IDF-2": ("TF-IDF", dict(ngram_range=(1, 2), min_df=2, sublinear_tf=False)),
        "TF-IDF-3": ("TF-IDF", dict(ngram_range=(1, 2), min_df=2, sublinear_tf=True)),
        "TF-IDF-4": ("TF-IDF", dict(ngram_range=(1, 2), min_df=1, sublinear_tf=True)),

        "FT-1": ("FastText", dict(vector_size=100, window=5, sg=1)),
        "FT-2": ("FastText", dict(vector_size=100, window=3, sg=1)),
        "FT-3": ("FastText", dict(vector_size=100, window=5, sg=0)),
        "FT-4": ("FastText", dict(vector_size=200, window=5, sg=1)),

        "MiniLM-1": ("MiniLM", {})
    }

    # 2. Shared parameters
    shared = {
        "BoW": dict(max_features=None, max_df=0.95),
        "TF-IDF": dict(max_features=None, max_df=0.95),
        "FastText": dict(
            min_count=2, epochs=20, min_n=3, max_n=6,
            negative=5, workers=1, seed=42
        ),
        "MiniLM": {}
    }

    functions = {
        "BoW": bag_of_words,
        "TF-IDF": tfidf_vectorization,
        "FastText": fasttext_vectorization,
        "MiniLM": minilm_vectorization
    }

    transformations = {}

    print("=" * 85)
    print("GENERATING FEATURE EXTRACTION CONFIGURATIONS")
    print("=" * 85)

    # 3. Generate feature matrices
    for name, (method, settings) in configs.items():
        params = {**shared[method], **settings}

        train, val, test, model = functions[method](
            train_df, val_df, test_df, **params
        )

        # Include all relevant parameters in the description
        if method == "MiniLM":
            info_params = {
                "model": "sentence-transformers/all-MiniLM-L6-v2",
                "device": str(model.device),
                **params
            }
        else:
            info_params = params

        info = (
            f"{method}: " +
            ", ".join(f"{k}={v}" for k, v in info_params.items())
        )

        transformations[name] = {
            "train": train,
            "val": val,
            "test": test,
            "info": info
        }

        print(f"\n{name}")
        print(f"  Train: {train.shape} | Val: {val.shape} | Test: {test.shape}")
        print(f"  {info}")

    print("\n" + "=" * 85)
    print(f"COMPLETED: {len(transformations)} configurations generated")
    print("=" * 85)

    return transformations
````

### 1. Support Vector Machine (Linear SVM)

````python
import pandas as pd
import time
from itertools import product
from sklearn.svm import LinearSVC, SVC
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import f1_score, accuracy_score


def evaluate_svm_models(transformations, train_df, val_df):
    """
    Evaluate 84 SVM configurations using 5-fold stratified CV.
    Rank by CV macro-F1 and retain fitted models.
    """

    features = [
        "BoW-1", "BoW-2", "BoW-3",
        "TF-IDF-1", "TF-IDF-2", "TF-IDF-3", "TF-IDF-4"
    ]

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    y_train = train_df["label"].to_numpy()
    y_val = val_df["label"].to_numpy()

    configs = []

    # LinearSVC: 3 C values x 2 class weights
    for C, weight in product([0.05, 0.1, 0.2, 0.5, 1], [None, "balanced"]):
        configs.append((
            LinearSVC(
                C=C,
                class_weight=weight,
                max_iter=10000,
                tol=1e-3,
                dual=True,
                random_state=42
            ),
            f"LinearSVC(C={C}, class_weight={weight!r}, max_iter=10000)"
        ))

    # SVC RBF: 3 C values x 2 gamma values
    #for C, gamma in product([0.1, 1, 10], ["scale", 0.01]):
    #    configs.append((
    #        SVC(
    #            C=C,
    #            gamma=gamma,
    #            kernel="rbf",
    #            class_weight=None
    #        ),
    #        f"SVC(C={C}, gamma={gamma!r}, kernel='rbf', class_weight=None)"
    #    ))

    results = {}
    summary = []

    total_experiments = len(features) * len(configs)
    total_start_time = time.time()

    print("=" * 70, flush=True)
    print("STARTING SVM MODEL EVALUATION", flush=True)
    print(f"Total experiments: {total_experiments}", flush=True)
    print("Cross-validation: 5 folds", flush=True)
    print("=" * 70, flush=True)

    for feature in features:
        X_train = transformations[feature]["train"]
        X_val = transformations[feature]["val"]

        print(f"\n{'=' * 70}", flush=True)
        print(f"FEATURE EXTRACTION: {feature}", flush=True)
        print(f"{'=' * 70}", flush=True)

        for model, description in configs:
            experiment_id = f"svm_{len(results) + 1}"
            experiment_number = len(results) + 1
            start_time = time.time()

            # PRINT 1: Starting experiment
            print(
                f"\n[{experiment_number}/{total_experiments}] "
                f"{experiment_id} | {feature} | {description}",
                flush=True
            )

            # 1. Cross-validation exclusively on training data
            print(
                "   [1/3] Running 5-fold cross-validation...",
                flush=True
            )

            cv_scores = cross_val_score(
                model,
                X_train,
                y_train,
                cv=cv,
                scoring="f1_macro",
                n_jobs=-1
            )

            print(
                f"   [1/3] CV completed | "
                f"Macro-F1: {cv_scores.mean():.4f} "
                f"± {cv_scores.std():.4f}",
                flush=True
            )

            # 2. Fit on complete training split
            print(
                "   [2/3] Training final model...",
                flush=True
            )

            model.fit(X_train, y_train)

            # 3. Evaluate on independent validation set
            print(
                "   [3/3] Evaluating validation set...",
                flush=True
            )

            predictions = model.predict(X_val)

            metrics = {
                "cv_macro_f1_mean": float(cv_scores.mean()),
                "cv_macro_f1_std": float(cv_scores.std()),
                "val_macro_f1": float(
                    f1_score(y_val, predictions, average="macro")
                ),
                "val_accuracy": float(
                    accuracy_score(y_val, predictions)
                )
            }

            # 4. Store experiment and fitted estimator
            results[experiment_id] = {
                "results": metrics,
                "info_input": feature,
                "info_modelo": description,
                "model": model
            }

            summary.append({
                "experiment": experiment_id,
                "feature": feature,
                "model": description,
                **metrics
            })

            elapsed = time.time() - start_time

            # PRINT 4: Experiment completed
            print(
                f"   ✓ COMPLETED in {elapsed:.1f}s | "
                f"CV F1: {metrics['cv_macro_f1_mean']:.4f} | "
                f"Val F1: {metrics['val_macro_f1']:.4f} | "
                f"Val Acc: {metrics['val_accuracy']:.4f}",
                flush=True
            )

    # 5. Rank experiments by cross-validation macro-F1
    summary_df = pd.DataFrame(summary).sort_values(
        "cv_macro_f1_mean",
        ascending=False
    ).reset_index(drop=True)

    total_elapsed = time.time() - total_start_time

    print("\n" + "=" * 70)
    print("SVM EVALUATION COMPLETED")
    print("=" * 70)
    print(f"Total experiments: {len(results)}")
    print(f"Total execution time: {total_elapsed / 60:.2f} minutes")

    print("\nSVM EXPERIMENT RANKING (BEST TO WORST)")
    print(
        summary_df.to_string(
            index=False,
            float_format="%.4f"
        )
    )

    # 6. Highlight best configuration
    best = summary_df.iloc[0]

    print("\n" + "=" * 70)
    print("BEST SVM CONFIGURATION")
    print("=" * 70)
    print(f"Experiment:        {best['experiment']}")
    print(f"Feature extraction:{best['feature']:>15}")
    print(f"Model:             {best['model']}")
    print(
        f"CV Macro-F1:       {best['cv_macro_f1_mean']:.4f} "
        f"± {best['cv_macro_f1_std']:.4f}"
    )
    print(f"Validation F1:     {best['val_macro_f1']:.4f}")
    print(f"Validation Acc:    {best['val_accuracy']:.4f}")

    return results
````

````python
import pandas as pd
import time
from sklearn.naive_bayes import MultinomialNB, ComplementNB, BernoulliNB
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import f1_score, accuracy_score


def evaluate_naive_bayes_models(transformations, train_df, val_df):
    """
    Evaluate 65 Naive Bayes configurations using stratified 5-fold CV.
    Rank by CV macro-F1 and retain fitted models.
    """

    features = [
        "BoW-1", "BoW-2", "BoW-3",
        "TF-IDF-1", "TF-IDF-2", "TF-IDF-3", "TF-IDF-4"
    ]

    cv = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )

    y_train = train_df["label"].to_numpy()
    y_val = val_df["label"].to_numpy()

    results_nb = {}
    summary = []

    total_experiments = 65
    total_start_time = time.time()

    print("=" * 70, flush=True)
    print("STARTING NAIVE BAYES MODEL EVALUATION", flush=True)
    print(f"Total experiments: {total_experiments}", flush=True)
    print("Cross-validation: 5 folds", flush=True)
    print("=" * 70, flush=True)

    for feature in features:
        X_train = transformations[feature]["train"]
        X_val = transformations[feature]["val"]

        print(f"\n{'=' * 70}", flush=True)
        print(f"FEATURE EXTRACTION: {feature}", flush=True)
        print(f"{'=' * 70}", flush=True)

        # Configure Naive Bayes models for this representation
        configs = [
            (
                MultinomialNB(alpha=a),
                f"MultinomialNB(alpha={a})"
            )
            for a in [0.01, 0.1, 0.5, 1, 5]
        ] + [
            (
                ComplementNB(alpha=a),
                f"ComplementNB(alpha={a})"
            )
            for a in [0.01, 0.1, 0.5, 1]
        ]

        # BernoulliNB only on binary Bag-of-Words
        if feature == "BoW-3":
            configs += [
                (
                    BernoulliNB(alpha=a, binarize=0.0),
                    f"BernoulliNB(alpha={a}, binarize=0.0)"
                )
                for a in [0.1, 1]
            ]

        for model, description in configs:
            experiment_number = len(results_nb) + 1
            experiment_id = f"nb_{experiment_number}"
            start_time = time.time()

            # PRINT 1: Starting experiment
            print(
                f"\n[{experiment_number}/{total_experiments}] "
                f"{experiment_id} | {feature} | {description}",
                flush=True
            )

            # 1. Cross-validation on training data only
            print(
                "   [1/3] Running 5-fold cross-validation...",
                flush=True
            )

            cv_scores = cross_val_score(
                model,
                X_train,
                y_train,
                cv=cv,
                scoring="f1_macro",
                n_jobs=-1
            )

            print(
                f"   [1/3] CV completed | "
                f"Macro-F1: {cv_scores.mean():.4f} "
                f"± {cv_scores.std():.4f}",
                flush=True
            )

            # 2. Fit on the complete training split
            print(
                "   [2/3] Training final model...",
                flush=True
            )

            model.fit(X_train, y_train)

            print(
                "   [2/3] Training completed.",
                flush=True
            )

            # 3. Evaluate on the independent validation set
            print(
                "   [3/3] Evaluating validation set...",
                flush=True
            )

            predictions = model.predict(X_val)

            metrics = {
                "cv_macro_f1_mean": float(cv_scores.mean()),
                "cv_macro_f1_std": float(cv_scores.std()),
                "val_macro_f1": float(
                    f1_score(y_val, predictions, average="macro")
                ),
                "val_accuracy": float(
                    accuracy_score(y_val, predictions)
                )
            }

            # 4. Store results and fitted estimator
            results_nb[experiment_id] = {
                "results": metrics,
                "info_input": feature,
                "info_modelo": description,
                "model": model
            }

            summary.append({
                "experiment": experiment_id,
                "feature": feature,
                "model": description,
                **metrics
            })

            elapsed = time.time() - start_time

            # PRINT 4: Experiment completed
            print(
                f"   ✓ COMPLETED in {elapsed:.1f}s | "
                f"CV F1: {metrics['cv_macro_f1_mean']:.4f} | "
                f"Val F1: {metrics['val_macro_f1']:.4f} | "
                f"Val Acc: {metrics['val_accuracy']:.4f}",
                flush=True
            )

            # PRINT 5: Overall progress
            progress = experiment_number / total_experiments * 100

            print(
                f"   Overall progress: "
                f"{experiment_number}/{total_experiments} "
                f"({progress:.1f}%)",
                flush=True
            )

    # 5. Rank by cross-validation macro-F1
    summary_df = pd.DataFrame(summary).sort_values(
        "cv_macro_f1_mean",
        ascending=False
    ).reset_index(drop=True)

    total_elapsed = time.time() - total_start_time

    print("\n" + "=" * 70, flush=True)
    print("NAIVE BAYES EVALUATION COMPLETED", flush=True)
    print("=" * 70, flush=True)
    print(f"Total experiments: {len(results_nb)}", flush=True)
    print(
        f"Total execution time: {total_elapsed / 60:.2f} minutes",
        flush=True
    )

    print("\nNAIVE BAYES EXPERIMENT RANKING (BEST TO WORST)")
    print(
        summary_df.to_string(
            index=False,
            float_format="%.4f"
        )
    )

    # 6. Highlight best configuration
    best = summary_df.iloc[0]

    print("\n" + "=" * 70)
    print("BEST NAIVE BAYES CONFIGURATION")
    print("=" * 70)
    print(f"Experiment:      {best['experiment']}")
    print(f"Feature:         {best['feature']}")
    print(f"Model:           {best['model']}")
    print(
        f"CV Macro-F1:     {best['cv_macro_f1_mean']:.4f} "
        f"± {best['cv_macro_f1_std']:.4f}"
    )
    print(f"Validation F1:   {best['val_macro_f1']:.4f}")
    print(f"Validation Acc:  {best['val_accuracy']:.4f}")

    return results_nb
````

### 3. MLP

````python
import pandas as pd
import time
from itertools import product
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import f1_score, accuracy_score
from scipy.sparse import issparse


def evaluate_mlp_models(transformations, train_df, val_df):
    """
    Evaluate 192 MLP configurations using stratified 5-fold CV.
    Supports sparse BoW/TF-IDF and dense MiniLM embeddings.
    Reports progress, execution times, and estimated remaining time.
    """

    features = ["TF-IDF-1", "BoW-1", "MiniLM-1"]

    cv = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )

    # Encode labels
    label_encoder = LabelEncoder()

    y_train = label_encoder.fit_transform(train_df["label"])
    y_val = label_encoder.transform(val_df["label"])

    # Hyperparameter grid
    grid = product(
        [(128,), (32,), (128, 64), (2,)],
        ["relu", "tanh"],
        [0.0001, 0.01],
        [0.0003, 0.001],
        [64, 128]
    )

    configs = list(grid)

    results_mlp = {}
    summary = []

    total = len(features) * len(configs)
    total_start_time = time.time()

    print("=" * 70, flush=True)
    print("STARTING MLP MODEL EVALUATION", flush=True)
    print(f"Total experiments: {total}", flush=True)
    print(f"Feature representations: {len(features)}", flush=True)
    print(f"Configurations per feature: {len(configs)}", flush=True)
    print("Cross-validation: 5 folds", flush=True)
    print(f"Total model fits: {total * 6}", flush=True)

    print("\nLABEL ENCODING", flush=True)
    for i, label in enumerate(label_encoder.classes_):
        print(f"   {label} -> {i}", flush=True)

    print("=" * 70, flush=True)

    for feature in features:
        X_train = transformations[feature]["train"]
        X_val = transformations[feature]["val"]

        feature_start_time = time.time()

        # Detect sparse matrices automatically
        is_sparse = issparse(X_train)

        print(f"\n{'=' * 70}", flush=True)
        print(f"FEATURE EXTRACTION: {feature}", flush=True)
        print(f"Train shape: {X_train.shape}", flush=True)
        print(f"Validation shape: {X_val.shape}", flush=True)
        print(f"Sparse representation: {is_sparse}", flush=True)
        print(f"Scaler with_mean: {not is_sparse}", flush=True)
        print(f"{'=' * 70}", flush=True)

        for hidden, activation, alpha, lr, batch in configs:
            experiment_number = len(results_mlp) + 1
            experiment_id = f"mlp_{experiment_number}"
            start_time = time.time()

            params = dict(
                hidden_layer_sizes=hidden,
                activation=activation,
                alpha=alpha,
                learning_rate_init=lr,
                batch_size=batch,
                early_stopping=True,
                max_iter=300,
                random_state=42,
                solver="adam",
                n_iter_no_change=10
            )

            model = Pipeline([
                ("scaler", StandardScaler(with_mean=not is_sparse)),
                ("mlp", MLPClassifier(**params))
            ])

            # PRINT 1: Starting experiment
            print(
                f"\n[{experiment_number}/{total}] "
                f"{experiment_id} | {feature}",
                flush=True
            )

            print(
                f"   Configuration: "
                f"hidden={hidden}, "
                f"activation={activation}, "
                f"alpha={alpha}, "
                f"lr={lr}, "
                f"batch={batch}",
                flush=True
            )

            # 1. Cross-validation
            print(
                "   [1/3] Running 5-fold cross-validation...",
                flush=True
            )

            cv_start_time = time.time()

            cv_scores = cross_val_score(
                model,
                X_train,
                y_train,
                cv=cv,
                scoring="f1_macro",
                n_jobs=1
            )

            cv_elapsed = time.time() - cv_start_time

            print(
                f"   [1/3] CV completed in {cv_elapsed:.1f}s | "
                f"Macro-F1: {cv_scores.mean():.4f} "
                f"± {cv_scores.std():.4f}",
                flush=True
            )

            # 2. Fit on complete training split
            print(
                "   [2/3] Training final model...",
                flush=True
            )

            fit_start_time = time.time()

            model.fit(X_train, y_train)

            fit_elapsed = time.time() - fit_start_time

            print(
                f"   [2/3] Training completed in {fit_elapsed:.1f}s",
                flush=True
            )

            # 3. Evaluate validation set
            print(
                "   [3/3] Evaluating validation set...",
                flush=True
            )

            val_start_time = time.time()

            predictions = model.predict(X_val)

            metrics = {
                "cv_macro_f1_mean": float(cv_scores.mean()),
                "cv_macro_f1_std": float(cv_scores.std()),
                "val_macro_f1": float(
                    f1_score(y_val, predictions, average="macro")
                ),
                "val_accuracy": float(
                    accuracy_score(y_val, predictions)
                )
            }

            val_elapsed = time.time() - val_start_time

            print(
                f"   [3/3] Validation completed in {val_elapsed:.1f}s",
                flush=True
            )

            description = (
                "MLPClassifier(" +
                ", ".join(f"{k}={v!r}" for k, v in params.items()) +
                ")"
            )

            # 4. Store results and fitted pipeline
            results_mlp[experiment_id] = {
                "results": metrics,
                "info_input": feature,
                "info_modelo": description,
                "model": model
            }

            summary.append({
                "experiment": experiment_id,
                "feature": feature,
                "model": description,
                **metrics
            })

            # 5. Progress reporting
            completed = len(results_mlp)
            elapsed = time.time() - start_time
            total_elapsed = time.time() - total_start_time

            progress = completed / total * 100
            avg_time = total_elapsed / completed
            remaining = (total - completed) * avg_time

            print(
                f"   ✓ COMPLETED in {elapsed:.1f}s | "
                f"CV F1: {metrics['cv_macro_f1_mean']:.4f} | "
                f"Val F1: {metrics['val_macro_f1']:.4f} | "
                f"Val Acc: {metrics['val_accuracy']:.4f}",
                flush=True
            )

            print(
                f"   Overall progress: "
                f"{completed}/{total} ({progress:.1f}%) | "
                f"Elapsed: {total_elapsed / 60:.1f} min | "
                f"ETA: {remaining / 60:.1f} min",
                flush=True
            )

        # Feature completion
        feature_elapsed = time.time() - feature_start_time

        print(f"\n{'-' * 70}", flush=True)
        print(
            f"FINISHED FEATURE: {feature} | "
            f"Time: {feature_elapsed / 60:.2f} minutes",
            flush=True
        )
        print(f"{'-' * 70}", flush=True)

    # 6. Rank by CV macro-F1
    summary_df = pd.DataFrame(summary).sort_values(
        "cv_macro_f1_mean",
        ascending=False
    ).reset_index(drop=True)

    total_elapsed = time.time() - total_start_time

    print("\n" + "=" * 70, flush=True)
    print("MLP EVALUATION COMPLETED", flush=True)
    print("=" * 70, flush=True)
    print(f"Total experiments: {len(results_mlp)}", flush=True)
    print(
        f"Total execution time: {total_elapsed / 60:.2f} minutes",
        flush=True
    )

    print("\nMLP EXPERIMENT RANKING (BEST TO WORST)")
    print(
        summary_df.to_string(
            index=False,
            float_format="%.4f"
        )
    )

    # 7. Highlight best configuration
    best = summary_df.iloc[0]

    print("\n" + "=" * 70)
    print("BEST MLP CONFIGURATION")
    print("=" * 70)
    print(f"Experiment:      {best['experiment']}")
    print(f"Feature:         {best['feature']}")
    print(f"Model:           {best['model']}")
    print(
        f"CV Macro-F1:     {best['cv_macro_f1_mean']:.4f} "
        f"± {best['cv_macro_f1_std']:.4f}"
    )
    print(f"Validation F1:   {best['val_macro_f1']:.4f}")
    print(f"Validation Acc:  {best['val_accuracy']:.4f}")

    return results_mlp
````

````python

import pandas as pd
from IPython.display import display


def display_top_models(results_list):
    """
    Combine experiment results and display the top 10 models
    ranked by cross-validation macro-F1.
    """
    rows = []

    for results_dict in results_list:
        for experiment_id, experiment in results_dict.items():
            rows.append({
                "Experiment": experiment_id,
                "Features": experiment["info_input"],
                "Model": experiment["info_modelo"],
                **experiment["results"]
            })

    columns = [
        "Experiment", "Features", "Model",
        "cv_macro_f1_mean", "cv_macro_f1_std",
        "val_macro_f1", "val_accuracy"
    ]

    top_10 = (
        pd.DataFrame(rows, columns=columns)
        .sort_values("cv_macro_f1_mean", ascending=False)
        .head(10)
        .reset_index(drop=True)
    )

    print(f"TOP {len(top_10)} MODELS — RANKED BY CV MACRO-F1")

    display(
        top_10.style.format({
            col: "{:.4f}" for col in columns[3:]
        }).highlight_max(
            subset=["cv_macro_f1_mean"], color="#d4edda"
        )
    )

    return top_10
````

````python

import pandas as pd


def generate_submission(results_svm, transformations, test_df, model_id="svm_36"):
    """
    Generate Kaggle submission using a fitted model.
    Save predictions to /kaggle/working/submission.csv.
    """
    valid_labels = {"anger", "disgust", "fear", "sadness", "surprise", "joy"}

    # 1. Retrieve fitted model and corresponding features
    experiment = results_svm[model_id]
    model = experiment["model"]
    info_input = experiment["info_input"]
    X_test = transformations[info_input]["test"]

    # 2. Generate predictions
    predictions = model.predict(X_test)

    # 3. Create submission preserving test row order
    submission = pd.DataFrame({
        "id": test_df["post_id"].to_numpy(),
        "label": predictions
    })

    # 4. Validate submission
    checks = {
        "Prediction count matches test set": len(predictions) == len(test_df),
        "Unique id values": submission["id"].is_unique,
        "No missing values": not submission.isna().any().any(),
        "All predicted labels are valid": submission["label"].isin(valid_labels).all()
    }

    print(f"Model: {model_id}")
    print(f"Features: {info_input}")
    print(f"Test observations: {len(test_df):,}")
    print(f"Predictions: {len(predictions):,}")

    print("\nSubmission validation:")
    for name, passed in checks.items():
        print(f"  {'OK' if passed else 'WARNING'}: {name}")

    # 5. Save only if all checks pass
    if all(checks.values()):
        path = "/kaggle/working/submission.csv"
        submission.to_csv(path, index=False)
        print(f"\nSubmission saved: {path}")
    else:
        print("\nWARNING: Submission not saved due to validation issues.")

    # 6. Display preview and class distribution
    print("\nSubmission preview:")
    display(submission.head())

    print("\nPredicted class distribution:")
    display(
        submission["label"]
        .value_counts()
        .rename_axis("Emotion")
        .reset_index(name="Count")
    )

    return submission
````

````python
import json
import re
import pandas as pd
import nltk

from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

# 1. Load data
BASE_PATH = "/kaggle/input/competitions/2026-dm-lab-2-private-competition/"

identification = pd.read_csv(BASE_PATH + "data_identification.csv")
emotions = pd.read_csv(BASE_PATH + "emotion.csv")

with open(BASE_PATH + "final_posts.json", encoding="utf-8") as f:
    posts = pd.DataFrame(json.load(f))

df = identification.merge(posts, on="post_id", how="left")
df = df.merge(emotions, on="post_id", how="left")

train_df = df[df["set"] == "train"].copy()
test_df = df[df["set"] == "test"].copy()

# 2. Remove duplicates
train_df = train_df.drop_duplicates(subset=["text", "label"]).reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

# 3. Text cleaning
def preprocessing_core(text):
    text = str(text).lower()
    text = re.sub(r"https?://\S+|www\.\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

train_df["text"] = train_df["text"].apply(preprocessing_core)
test_df["text"] = test_df["text"].apply(preprocessing_core)

# 4. Tokenization + lemmatization
for resource in ["punkt", "punkt_tab", "wordnet", "averaged_perceptron_tagger_eng"]:
    nltk.download(resource, quiet=True)

lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):
    return {
        "J": wordnet.ADJ,
        "V": wordnet.VERB,
        "N": wordnet.NOUN,
        "R": wordnet.ADV
    }.get(tag[0], wordnet.NOUN)

def tokenize_lemma(text):
    tokens = word_tokenize(text)
    tagged = pos_tag(tokens)
    return [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in tagged]

train_df["tokens_lemma"] = train_df["text"].apply(tokenize_lemma)
test_df["tokens_lemma"] = test_df["text"].apply(tokenize_lemma)

# 5. TF-IDF-1
config = {
    "TF-IDF-1": (
        "TF-IDF",
        dict(ngram_range=(1, 1), min_df=2, sublinear_tf=False)
    )
}

vectorizer = TfidfVectorizer(
    **config["TF-IDF-1"][1],
    max_df=0.95,
    max_features=None,
    analyzer=lambda tokens: tokens,
    lowercase=False,
    token_pattern=None
)

X_train = vectorizer.fit_transform(train_df["tokens_lemma"])
X_test = vectorizer.transform(test_df["tokens_lemma"])

# 6. Train final SVM on all training data
model = LinearSVC(
    C=0.2,
    class_weight="balanced",
    max_iter=10000,
    random_state=42
)

model.fit(X_train, train_df["label"])

# 7. Generate submission
predictions = model.predict(X_test)

submission = pd.DataFrame({
    "id": test_df["post_id"].values,
    "label": predictions
})

submission.to_csv("/kaggle/working/submission.csv", index=False)

# 8. Final output
print("submission.csv generated successfully")
````
